```python
import pandas as pd
```

## 1. Criando o DataFrame
Cada linha é um produto; as colunas são categoria, nome, quantidade em estoque e preço unitário (R$).

```python
produtos = pd.DataFrame(
    [
        ['Notebook',   'Notebook i5 15"',       12, 3499.90],
        ['Periférico', 'Mouse sem fio',          85,   79.90],
        ['Periférico', 'Teclado mecânico',       30,  289.00],
        ['Monitor',    'Monitor 24" Full HD',    18,  899.00],
        ['Armazenamento', 'SSD 1 TB',            40,  429.90],
        ['Notebook',   'Notebook i7 14"',         6, 5299.00],
        ['Armazenamento', 'Pen drive 64 GB',    120,   39.90],
    ],
    columns=['Categoria', 'Produto', 'Estoque', 'Preço']
)
print(produtos)
```

## 3. Tipos de dados
`Categoria` tem poucos valores que se repetem, então pode ser convertida para `category`, que ocupa menos memória e deixa claro que é uma variável categórica.

```python
print(produtos.dtypes)
produtos['Categoria'] = produtos['Categoria'].astype('category')
print('\nDepois da conversão:')
print(produtos.dtypes)
print('\nCategorias:', list(produtos['Categoria'].cat.categories))
```

## 4. Adicionando e removendo colunas

```python
# Nova coluna calculada: valor total parado em estoque
produtos['Valor_em_estoque'] = produtos['Estoque'] * produtos['Preço']

# Nova coluna condicional: precisa repor? (regra fictícia: menos de 20 unidades)
produtos['Repor'] = produtos['Estoque'] < 20

produtos
```

```python
# Coluna temporária, só para mostrar a remoção
produtos['Observacao'] = 'conferido'
produtos = produtos.drop(columns=['Observacao'])
produtos.columns
```

## 5. Operações básicas

```python
print('Valor total do estoque: R$', round(produtos['Valor_em_estoque'].sum(), 2))
print('Produtos para repor:')
print(produtos.loc[produtos['Repor'], ['Produto', 'Estoque']])
```

```python
produtos.sort_values('Valor_em_estoque', ascending=False)[['Produto', 'Valor_em_estoque']]
```

```python
produtos.groupby('Categoria', observed=True)['Valor_em_estoque'].sum().sort_values(ascending=False)
```

**Leitura:** os notebooks são só 18 das 311 unidades em estoque, mas concentram a maior parte do dinheiro parado. Ao todo são R$ 127.420,30 em estoque, e só os notebooks somam R$ 73.792,80 (cerca de 58%). O pen drive é o item com mais unidades (120) e o que menos pesa no valor total (R$ 4.788,00).

## 6. Salvando e lendo o DataFrame

```python
produtos.to_csv('estoque_loja.csv', index=False)
conferencia = pd.read_csv('estoque_loja.csv')
print(conferencia.shape)
conferencia.head(3)
```

Ao ler o CSV de volta, `Categoria` volta como texto (`object`/`str`): o formato CSV não guarda o tipo `category`, então a conversão precisa ser refeita depois de carregar o arquivo.

```python
import pandas as pd
from io import StringIO

# Dados do dataset
dados = """id_chamado,data_abertura,setor,categoria,prioridade,status
CH001,2026-08-20,Financeiro,Acesso,Alta,Resolvido
CH002,2026-08-20,Acadêmico,Sistema lento,Média,Resolvido
CH003,2026-08-21,Biblioteca,Impressora,Baixa,Resolvido
CH004,2026-08-21,Financeiro,Erro de sistema,Crítica,Resolvido
CH005,2026-08-22,RH,Acesso,Média,Em andamento
CH006,2026-08-22,Acadêmico,Rede,Alta,Resolvido
CH007,2026-08-23,Laboratório,Hardware,Média,Resolvido
CH008,2026-08-23,Biblioteca,Sistema lento,Baixa,Resolvido
CH009,2026-08-24,RH,Erro de sistema,Alta,Resolvido
CH010,2026-08-24,Laboratório,Rede,Crítica,Resolvido
CH011,2026-08-25,Financeiro,Hardware,Média,Aberto
CH012,2026-08-25,Acadêmico,Acesso,Baixa,Resolvido
CH013,2026-08-26,Biblioteca,Rede,Média,Resolvido
CH014,2026-08-26,RH,Sistema lento,Alta,Resolvido
CH015,2026-08-27,Laboratório,Erro de sistema,Crítica,Resolvido
CH016,2026-08-27,Financeiro,Acesso,Baixa,Resolvido
CH017,2026-08-28,Acadêmico,Hardware,Média,Resolvido
CH018,2026-08-28,Laboratório,Sistema lento,Média,Resolvido"""

# Criar o DataFrame
df = pd.read_csv(StringIO(dados))

# 5 primeiras linhas
print("5 primeiras linhas:")
print(df.head())

# 3 últimas linhas
print("\n3 últimas linhas:")
print(df.tail(3))

# Dimensões do DataFrame
print("\nDimensões:")
print(df.shape)

# Nomes das colunas
print("\nNomes das colunas:")
print(df.columns)
```

```python
# Verificar informações do DataFrame
print("Informações do DataFrame:")
df.info()

# Estatísticas descritivas
print("\nResumo estatístico:")
print(df.describe())

# Verificar valores ausentes
print("\nValores ausentes por coluna:")
print(df.isnull().sum())
```

```python
# 1. Selecionar as colunas: id, setor, prioridade e status
selecao = df[["id_chamado", "setor", "prioridade", "status"]]

print("Seleção das colunas:")
print(selecao)


# 2. Com posições, mostrar 5 linhas e 4 colunas
recorte = df.iloc[:5, :4]

print("\nRecorte com 5 linhas e 4 colunas:")
print(recorte)
```

```python
# 1. Chamados com prioridade Alta
filtro_alta = df[df["prioridade"] == "Alta"]

print("Chamados com prioridade Alta:")
print(filtro_alta)
print("Quantidade:", filtro_alta.shape[0])


# 2. Chamados resolvidos acima de 24 horas
# OBS: O dataset fornecido não possui uma coluna de tempo de atendimento.
# Portanto, não é possível realizar esse filtro com os dados atuais.

# 3. Chamados das categorias Rede ou Acesso
filtro_rede_acesso = df[df["categoria"].isin(["Rede", "Acesso"])]

print("\nChamados de Rede ou Acesso:")
print(filtro_rede_acesso)
print("Quantidade:", filtro_rede_acesso.shape[0])
```

```python
# Converter a coluna de data para o formato datetime
df["data_abertura"] = pd.to_datetime(df["data_abertura"])

# Criar o tempo em dias
# Neste caso, usando a diferença entre a data atual e a data de abertura
data_atual = pd.Timestamp("2026-08-29")
df["tempo_dias"] = (data_atual - df["data_abertura"]).dt.days

# Calcular o tempo médio por categoria
media_categoria = (
    df.groupby("categoria")["tempo_dias"]
    .mean()
    .sort_values(ascending=False)
)

print("Tempo médio por categoria:")
print(media_categoria)
```
